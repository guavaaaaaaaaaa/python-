```python
import pandas as pd
import numpy as np
import joblib
import tensorflow as tf
from tensorflow import keras
from sklearn.preprocessing import StandardScaler
import os

# 1. 設定路徑與載入資料
input_file = '/content/drive/MyDrive/緯育/團專-紅白酒/01-資料/test/test_group_red.csv'
model_path = '/content/drive/MyDrive/緯育/團專-紅白酒/04-模型定稿/red_MLP_final.keras'
scaler_path = '/content/drive/MyDrive/緯育/團專-紅白酒/04-模型定稿/red_scaler.gz'
output_file = '/content/drive/MyDrive/緯育/團專-紅白酒/predictions_red.csv'

df = pd.read_csv(input_file)

# 2. 載入模型與 Scaler 工具
model = keras.models.load_model(model_path)
scaler = joblib.load(scaler_path)

# 3. 特徵提取與處理
TARGET_COL = "quality"
# 確保排除掉 target 和 ID，只留下模型需要的特徵
X_raw = df.drop(columns=[TARGET_COL, "row_id"], errors='ignore').values.astype(np.float32)

# 💡 關鍵修正 1：使用 Scaler 轉換資料
# 模型需要的是縮放後的數值，不是 scaler 物件本身
X_scaled = scaler.transform(X_raw)

# 4. 執行預測 (得到 [樣本數, 3] 的機率矩陣)
print(f"🚀 正在針對 {len(df)} 筆資料進行預測...")
raw_predictions = model.predict(X_scaled, batch_size=128, verbose=1)

# 💡 關鍵修正 2：針對 Dense(3) 使用 argmax 取出類別
# 取出 3 個輸出中機率最大的索引 (0, 1, 或 2)
predictions = np.argmax(raw_predictions, axis=1)

# 5. 將預測結果存入 DataFrame
df['predicted_quality'] = predictions

# 6. 存檔與檢查
df.to_csv(output_file, index=False)

print("-" * 30)
print(f"✅ 預測完成！檔案已存至：{output_file}")
print(f"原始資料筆數: {len(df)}")
print(f"預測標籤筆數: {len(predictions)}")
print("\n前五筆預測結果：")
print(df[['quality', 'predicted_quality']].head())

print("\n預測類別分佈統計：")
print(df['predicted_quality'].value_counts().sort_index())
```
